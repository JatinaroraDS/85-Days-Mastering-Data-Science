```python
# When calculating statistics from a frequency table, the ages must be multiplied by their respective frequency to find the true weighted total. Here is the verified breakdown of the data:

#• Total Frequency (Count): 45 (13 + 12 + 20)
#• Total Weighted Age: 1,261 [(25 × 13) + (28 × 12) + (30 × 20)]
#• Exact Mean (Average): 28.02
# The exact mean is 28.02, which rounds cleanly to 28 if you are looking for the nearest whole number.

#python
ages = [25,28,30]
freqs = [13,12,20]
total_freqs = sum(freqs)
total_age_weighted = sum(a * f for a, f in zip(ages, freqs))
mean_age = total_age_weighted/total_freqs
print(f"total age: {total_age_weighted}")
print(f"mean age: {mean_age}")
print(f"total population: {total_freqs}")



```
