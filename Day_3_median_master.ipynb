```python
#Python
import pandas as pd

#--------------------------------------------------------------------
# PROBLEM 1 : upGrad Student Ages Dataset
#--------------------------------------------------------------------
# Inputting the exact 11 values from your Google Sheet column
ages_dataset = [12,13,17,14,16,18,21,15,18,17,19]

#Convert to a Pandas Series
ages_series = pd.Series(ages_dataset)

#Calculate statistics
mean_ages = ages_series.mean()
median_ages = ages_series.median()

print("---Problem 1 : Student Ages Verification---")
print(f"Excel Sheet Mean : 16.4 | Python Calculated: {mean_ages:.1f}")
print(f"Excel Sheet Median : 17.0 | Python Calculated {median_ages:.1f}")
print("--------------------------------------")
```

```python
#Python
ages = [12,13,17,14,16,18,21,15,18,17,19]

# 1. This single line tells Python to sort the data automatically
sorted_ages = sorted(ages)

# 2. Print it to see the ascending order instantly
print("Sorted Array:" , sorted_ages)

# 3. Find the median using pure Python math
n = len(sorted_ages)
median = sorted_ages[n//2]
print("Median:" , median)
```

```python
#python
import pandas as pd
#-----------------------------------------------------------------
#PROBLEM 2 : Employee Performance Ratings (Even Dataset)
#-----------------------------------------------------------------
#1. Inputting the exact raw values from your Google Sheet columns
ratings_data = [6, 8, 9, 7, 10, 10, 5, 5]

#Convert the list to a Pandas Series
ratings_series = pd.Series(ratings_data)

#2. Automated Line of code to sort your data table in descending order
descending_ratings = ratings_series.sort_values(ascending=False).tolist()

#3.Compute central tendency metrics using Pandas shortcuts
mean_ratings = rating_series.mean()
pandas_median = rating_series.median()


print("==================================================")
print("   DAY 3: PROBLEM 2 PERFORMANCE RATINGS          ")
print("==================================================")
print(f"📋 Raw Input Ratings     : {ratings_data}")
print(f"📉 Automated Descending  : {descending_ratings}")
print(f"📊 Calculated Mean       : {mean_ratings:.2f}")
print(f"📌 Calculated Median     : {pandas_median} | (Excel Verified: 7.5)")
print("==================================================")


```
