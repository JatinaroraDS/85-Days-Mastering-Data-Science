```
import pandas as pd

# 1. Creating a dataset with the data types you studied today
data = {
    "City": ["Delhi", "Mumbai", "Bangalore"],      # Qualitative / Nominal Data
    "Rank": [1,2,3],                     # Quantitative / Ordinal Data (Fixed!)
    "Temperature_Celsius": [32.5, 28.0, 24.5]      # Quantitative / Continuous Data
}

# 2. Convert it into a structured DataFrame table
df = pd.DataFrame(data)

# 3. Check how Python reads these data types automatically
print(df.dtypes)
```
