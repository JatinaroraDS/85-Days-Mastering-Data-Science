```python
#===================================================================
# DAY 5: BOTTOM=UP MARKET SIZING & REVENUE OPTIMIZATION (F&B CASE STUDY)
#PUBLIC Repository Progress: https://github.com
import pandas as pd

print("===========================================")
print("DAY 5: F&B BUSINESS BOTTOM-UP MARKET SIZING")
print("===========================================")

# 1. Define the current operational metrics from the local F&B business
# Current active local customer pool across Cafe, Bar, and B&B
current_som_customers = 1200
average_spend_per_customer_inr= 850

# 2. Model the business revenue using a Bottom-Up Approach
current_annual_revenue = current_som_customers * average_spend_per_customer_inr * 12
print(f"Current Estimated Annual Revenue: ₹{current_annual_revenue:,.2f}")

# 3. Scenario: The owner wants to scale, but is afraid of adding ₹20,000 staff costs.
# Solution: Use data analytics to increase average customer spend by 15% via menu optimization.
optimized_spend_per_customer = average_spend_per_customer_inr * 1.15
scaled_annual_revenue = current_som_customers * optimized_spend_per_customer * 12

revenue_delta = scaled_annual_revenue - current_annual_revenue
staffing_cost_saved = 20000*12 #Cost of 1 additional full-time staff member per year

print("------------------------------------------------")
print("🚀 DATA-DRIVEN SCALING METRICS (WITHOUT EXTRA STAFF):")
print(f"🔹 Scaled Revenue (with 15% optimized spend): ₹{scaled_annual_revenue:,.2f}")
print(f"📈 Net Revenue Increase via Data Insights   : ₹{revenue_delta:,.2f}")
print(f"🛑 Traditional Staffing Overhead Avoided   : ₹{staffing_cost_saved:,.2f}")
print("==================================================")
print("💡 Operational Takeaway:")
print("The owner doesn't need more ₹20k staff to scale up operations.")
print("By running local Python analytics on transaction logs, they can identify")
print("cross-selling patterns to boost average spend safely within their current footprint.")
print("==================================================")
```
