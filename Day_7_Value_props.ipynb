```python
#=================================================================
# DAY 7: MARKETING VALUE PROPOSITION MAPPING & INTERACTIVE QUERY ENGINE
# Public Repository Progress: https://github.com
#=================================================================
import pandas as pd

print("=========================================================")
print("     DAY 7: CORPORATE VALUE PROPOSITION AUTOMATION       ")
print("=========================================================")


# 1. Define the structural matrix matching my upGrad module challenge
marketing_data = {
    "Value_Proposition": [
        "Anyone, anywhere, can start a business",
        "Great writing, simplified",
        "Unlock the power of video"
    ],
    "Target_product" : ["Shopify", "Grammarly", "Vimeo"]
}

df = pd.DataFrame(marketing_data)

# 2. Design the search automation function
def lookup_proposition(search_text):
    # Strip accidental leading/trailing spaces from user input text
    cleaned_text = search_text.strip()
    match = df[df["Value_Proposition"] == cleaned_text]
    if not match.empty:
        return f"Maps to --> {match['Target_product'].values} ✅"
    return " ❌ Error: Slogan not found in corporate records."

# 3. Interactive Input/Output Portal
print(" 📋 Live Database Reference Matrix:")
print(df.to_string(index=False))
print("----------------------------------------------------------")

# This line forces the terminal to wait for your live keyboard input
user_query = input("👉 Enter the exact Value Proposition slogan to audit: ")

#Generate and print the output instantly
search_result = lookup_proposition(user_query)
print("\n====================SYSTEM ENGINE OUTPUT================")
print(f"📥 INPUT SLOGAN: '{user_query}'")
print(f"📥 Search Result : {search_result}")
print("===========================================================")
```
